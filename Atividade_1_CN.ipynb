```python
import math
import numpy as np
import matplotlib.pyplot as plt
# Define um estilo visual claro para todos os gráficos do notebook.
plt.style.use("seaborn-v0_8-whitegrid")
```

```python
x1 = 3.14159265
x2 = 98.76543
x3 = 0.00498765

n = [0, 1, 2, 3, 4]

```

```python
# Arredontamento

x1arre = [round(x1, i) for i in n]
x2arre = [round(x2, i) for i in n]
x3arre = [round(x3, i) for i in n]

print(x1arre)
print(x2arre)
print(x3arre)
```

```python
# Truncamento

x1trunc = [(math.trunc((10**i)*x1))/(10**i) for i in n]
x2trunc = [(math.trunc((10**i)*x2))/(10**i) for i in n]
x3trunc = [(math.trunc((10**i)*x3))/(10**i) for i in n]

print(x1trunc)
print(x2trunc)
print(x3trunc)
```

```python
def Eabs(x, xn):
  return abs(x - xn)

def Erel(x, xn):
  return (abs(x - xn)/abs(x))

def Eper(x, xn):
  return 100*Erel(x, xn)
```

```python
# Erros absolutos, relativos e percentuais para x1 Arredondado
ErrAbsX1 = [Eabs(x1, xn) for xn in x1arre]
print(f"Erros absolutos x1: {ErrAbsX1}")

ErrRelX1 = [Erel(x1, xn) for xn in x1arre]
print(f"Erros Relativos x1: {ErrRelX1}")

ErrPerX1 = [Eper(x1, xn) for xn in x1arre]
print(f"Erros Percentagens x1: {ErrPerX1}")
```

```python
# Erros absolutos, relativos e percentuais para x2 Arredondado
ErrAbsX2 = [Eabs(x2, xn) for xn in x2arre]
print(f"Erros absolutos x2: {ErrAbsX2}")

ErrRelX2 = [Erel(x2, xn) for xn in x2arre]
print(f"Erros Relativos x2: {ErrRelX2}")

ErrPerX2 = [Eper(x2, xn) for xn in x2arre]
print(f"Erros Percentagens x2: {ErrPerX2}")
```

```python
# Erros absolutos, relativos e percentuais para x3 Arredondado

ErrAbsX3 = [Eabs(x3, xn) for xn in x3arre]
print(f"Erros absolutos x3: {ErrAbsX3}")

ErrRelX3 = [Erel(x3, xn) for xn in x3arre]
print(f"Erros Relativos x3: {ErrRelX3}")

ErrPerX3 = [Eper(x3, xn) for xn in x3arre]
print(f"Erros Percentagens x3: {ErrPerX3}")
```

```python
# Erros absolutos, relativos e percentuais para x1 Truncado
ErrAbsX1Trunc = [Eabs(x1, xn) for xn in x1trunc]
print(f"Erros absolutos x1Trunc: {ErrAbsX1Trunc}")

ErrRelX1Trunc = [Erel(x1, xn) for xn in x1trunc]
print(f"Erros Relativos x1 TRrunc: {ErrRelX1Trunc}")

ErrPerX1Trunc = [Eper(x1, xn) for xn in x1trunc]
print(f"Erros Percentagens x1 Trunc: {ErrPerX1Trunc}")
```

```python
# Erros absolutos, relativos e percentuais para x2 Truncado
ErrAbsX2Trunc = [Eabs(x2, xn) for xn in x2trunc]
print(f"Erros absolutos x2 Trunc: {ErrAbsX2Trunc}")

ErrRelX2Trunc = [Erel(x2, xn) for xn in x2trunc]
print(f"Erros Relativos x2 Trunc: {ErrRelX2Trunc}")

ErrPerX2Trunc = [Eper(x2, xn) for xn in x2trunc]
print(f"Erros Percentagens x2 Trunc: {ErrPerX2Trunc}")
```

```python
# Erros absolutos, relativos e percentuais para x3 Truncado
ErrAbsX3Trunc = [Eabs(x3, xn) for xn in x3trunc]
print(f"Erros absolutos x3 Trunc: {ErrAbsX3Trunc}")

ErrRelX3Trunc = [Erel(x3, xn) for xn in x3trunc]
print(f"Erros Relativos x3 Trunc: {ErrRelX3Trunc}")

ErrPerX3Trunc = [Eper(x3, xn) for xn in x3trunc]
print(f"Erros Percentagens x3 Trunc: {ErrPerX3Trunc}")
```

```python
import pandas as pd

# Re-calculate errors for rounded values to ensure consistency
ErrAbsX1 = [Eabs(x1, xn) for xn in x1arre]
ErrRelX1 = [Erel(x1, xn) for xn in x1arre]
ErrPerX1 = [Eper(x1, xn) for xn in x1arre]

ErrAbsX2 = [Eabs(x2, xn) for xn in x2arre]
ErrRelX2 = [Erel(x2, xn) for xn in x2arre]
ErrPerX2 = [Eper(x2, xn) for xn in x2arre]

ErrAbsX3 = [Eabs(x3, xn) for xn in x3arre]
ErrRelX3 = [Erel(x3, xn) for xn in x3arre]
ErrPerX3 = [Eper(x3, xn) for xn in x3arre]

# Create DataFrame for x1 (Arredondamento)
data_x1_arre = {
    'Precisão (n)': n,
    'Valor Original': [x1] * len(n),
    'Valor Arredondado': x1arre,
    'Erro Absoluto': ErrAbsX1,
    'Erro Relativo': ErrRelX1,
    'Erro Percentual': ErrPerX1
}
df_x1_arre = pd.DataFrame(data_x1_arre)
print("Tabela de Erros para x1 (Arredondamento):")
print(df_x1_arre.to_string(index=False))
print("\n")

# Create DataFrame for x2 (Arredondamento)
data_x2_arre = {
    'Precisão (n)': n,
    'Valor Original': [x2] * len(n),
    'Valor Arredondado': x2arre,
    'Erro Absoluto': ErrAbsX2,
    'Erro Relativo': ErrRelX2,
    'Erro Percentual': ErrPerX2
}
df_x2_arre = pd.DataFrame(data_x2_arre)
print("Tabela de Erros para x2 (Arredondamento):")
print(df_x2_arre.to_string(index=False))
print("\n")

# Create DataFrame for x3 (Arredondamento)
data_x3_arre = {
    'Precisão (n)': n,
    'Valor Original': [x3] * len(n),
    'Valor Arredondado': x3arre,
    'Erro Absoluto': ErrAbsX3,
    'Erro Relativo': ErrRelX3,
    'Erro Percentual': ErrPerX3
}
df_x3_arre = pd.DataFrame(data_x3_arre)
print("Tabela de Erros para x3 (Arredondamento):")
print(df_x3_arre.to_string(index=False))
print("\n")


# --- Adicionando tabelas para valores truncados ---

# Create DataFrame for x1 (Truncamento)
data_x1_trunc = {
    'Precisão (n)': n,
    'Valor Original': [x1] * len(n),
    'Valor Truncado': x1trunc,
    'Erro Absoluto': ErrAbsX1Trunc,
    'Erro Relativo': ErrRelX1Trunc,
    'Erro Percentual': ErrPerX1Trunc
}
df_x1_trunc = pd.DataFrame(data_x1_trunc)
print("Tabela de Erros para x1 (Truncamento):")
print(df_x1_trunc.to_string(index=False))
print("\n")

# Create DataFrame for x2 (Truncamento)
data_x2_trunc = {
    'Precisão (n)': n,
    'Valor Original': [x2] * len(n),
    'Valor Truncado': x2trunc,
    'Erro Absoluto': ErrAbsX2Trunc,
    'Erro Relativo': ErrRelX2Trunc,
    'Erro Percentual': ErrPerX2Trunc
}
df_x2_trunc = pd.DataFrame(data_x2_trunc)
print("Tabela de Erros para x2 (Truncamento):")
print(df_x2_trunc.to_string(index=False))
print("\n")

# Create DataFrame for x3 (Truncamento)
data_x3_trunc = {
    'Precisão (n)': n,
    'Valor Original': [x3] * len(n),
    'Valor Truncado': x3trunc,
    'Erro Absoluto': ErrAbsX3Trunc,
    'Erro Relativo': ErrRelX3Trunc,
    'Erro Percentual': ErrPerX3Trunc
}
df_x3_trunc = pd.DataFrame(data_x3_trunc)
print("Tabela de Erros para x3 (Truncamento):")
print(df_x3_trunc.to_string(index=False))
print("\n")
```

```python
plt.figure(figsize=(15, 5))

# Plot para x1
plt.subplot(1, 3, 1) # 1 linha, 3 colunas, 1º gráfico
plt.plot(n, ErrAbsX1, marker='o', label='x1 Arredondado')
plt.plot(n, ErrAbsX1Trunc, marker='x', label='x1 Truncado')
plt.yscale('log')
plt.xlabel('Precisão (n)')
plt.ylabel('Erro Absoluto (log scale)')
plt.title('Erro Absoluto vs n para x1')
plt.legend()
plt.grid(True, which="both", ls="-", axis='y')

# Plot para x2
plt.subplot(1, 3, 2) # 1 linha, 3 colunas, 2º gráfico
plt.plot(n, ErrAbsX2, marker='o', label='x2 Arredondado')
plt.plot(n, ErrAbsX2Trunc, marker='x', label='x2 Truncado')
plt.yscale('log')
plt.xlabel('Precisão (n)')
plt.ylabel('Erro Absoluto (log scale)')
plt.title('Erro Absoluto vs n para x2')
plt.legend()
plt.grid(True, which="both", ls="-", axis='y')

# Plot para x3
plt.subplot(1, 3, 3) # 1 linha, 3 colunas, 3º gráfico
plt.plot(n, ErrAbsX3, marker='o', label='x3 Arredondado')
plt.plot(n, ErrAbsX3Trunc, marker='x', label='x3 Truncado')
plt.yscale('log')
plt.xlabel('Precisão (n)')
plt.ylabel('Erro Absoluto (log scale)')
plt.title('Erro Absoluto vs n para x3')
plt.legend()
plt.grid(True, which="both", ls="-", axis='y')

plt.tight_layout() # Ajusta o layout para evitar sobreposição
plt.show()
```

Sim, podemos afirmar que o arredondamento geraum erro menor ou igual ao truncamento, de acordo com os gráficos baseados nos dados calculado, pode-se perceber que a curva de arredondamento nunca está acima da de truncamento.

# Parte B: Medidas de pressão arterial
121,7; 120,4; 122,1; 119,8; 121,2; 120,9; 122,4; 121,0; 120,2; 121,5.

```python
medidasOriginais = [121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5]

medidasArredondadas = [round(medida, 0) for medida in medidasOriginais]

# math.trunc() já retorna a parte inteira. Multiplicar por 1 e dividir por 1 é redundante.
medidasTruncadas = [math.trunc(medida) for medida in medidasOriginais]

print(medidasOriginais)
print(medidasArredondadas)
print(medidasTruncadas)

media = np.mean(medidasOriginais)

print(f"\nMédia: {media}")
```

```python
# Funções auxiliares para estatísticas
def calculate_stats(data, name):
    mean_val = np.mean(data)
    min_val = np.min(data)
    max_val = np.max(data)
    amplitude = max_val - min_val
    std_dev = np.std(data, ddof=1) # ddof=1 para desvio-padrão amostral
    print(f"\n--- Estatísticas para {name} ---")
    print(f"Média: {mean_val:.4f}")
    print(f"Mínimo: {min_val:.4f}")
    print(f"Máximo: {max_val:.4f}")
    print(f"Amplitude: {amplitude:.4f}")
    print(f"Desvio-padrão amostral: {std_dev:.4f}")
    return mean_val

# 1. Estatísticas para os dados originais
mean_original = calculate_stats(medidasOriginais, "Dados Originais")

# 2. Estatísticas para os dados arredondados
mean_arredondadas = calculate_stats(medidasArredondadas, "Dados Arredondados")

# 3. Estatísticas para os dados truncados
mean_truncadas = calculate_stats(medidasTruncadas, "Dados Truncados")

print("\n--- Erros da Média Arredondada em relação à Média Original ---")
print(f"Média de Referência (Original): {mean_original:.4f}")
print(f"Média Arredondada: {mean_arredondadas:.4f}")
print(f"Erro Absoluto da Média Arredondada: {Eabs(mean_original, mean_arredondadas):.6f}")
print(f"Erro Relativo da Média Arredondada: {Erel(mean_original, mean_arredondadas):.6f}")
print(f"Erro Percentual da Média Arredondada: {Eper(mean_original, mean_arredondadas):.6f}%")

print("\n--- Erros da Média Truncada em relação à Média Original ---")
print(f"Média de Referência (Original): {mean_original:.4f}")
print(f"Média Truncada: {mean_truncadas:.4f}")
print(f"Erro Absoluto da Média Truncada: {Eabs(mean_original, mean_truncadas):.6f}")
print(f"Erro Relativo da Média Truncada: {Erel(mean_original, mean_truncadas):.6f}")
print(f"Erro Percentual da Média Truncada: {Eper(mean_original, mean_truncadas):.6f}%")
```
