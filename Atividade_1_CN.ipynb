```python
import math
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib import animation
from matplotlib.patches import Circle
from matplotlib.gridspec import GridSpec

# Define um estilo visual claro para todos os gráficos do notebook.
plt.style.use("seaborn-v0_8-whitegrid")

# Função para calcular o erro absoluto entre um valor de referência (x) e um valor aproximado (xn)
def Eabs(x, xn):
  return abs(x - xn)

# Função para calcular o erro relativo entre um valor de referência (x) e um valor aproximado (xn)
# Retorna infinito se x for zero, a menos que xn também seja zero (erro 0)
def Erel(x, xn):
  if x == 0:
    return 0 if xn == 0 else float('inf')
  return (abs(x - xn)/abs(x))

# Função para calcular o erro percentual a partir do erro relativo
def Eper(x, xn):
  return 100 * Erel(x, xn)

# Função auxiliar para truncar um número para 'i' casas decimais
def trunc_decimal(number, decimals=0):
    # Multiplica o número por 10 elevado ao número de casas decimais desejadas,
    # trunca o resultado (remove a parte fracionária) e depois divide pelo mesmo fator.
    if decimals < 0: # Não permite truncar para casas antes da vírgula desta forma
        raise ValueError("Decimais deve ser um número não negativo.")
    factor = 10**decimals
    return math.trunc(number * factor) / factor
```

# Parte B: Medidas de pressão arterial
121,7; 120,4; 122,1; 119,8; 121,2; 120,9; 122,4; 121,0; 120,2; 121,5.

```python
medidasOriginais = [121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5]

medidasArredondadas = [round(medida, 0) for medida in medidasOriginais]

# math.trunc() já retorna a parte inteira. Multiplicar por 1 e dividir por 1 é redundante.
medidasTruncadas = [math.trunc(medida) for medida in medidasOriginais]

print(medidasOriginais)
print(medidasArredondadas)
print(medidasTruncadas)

media = np.mean(medidasOriginais)

print(f"\nMédia: {media}")
```

```python
# Funções auxiliares para estatísticas
def calculate_stats(data, name):
    mean_val = np.mean(data)
    min_val = np.min(data)
    max_val = np.max(data)
    amplitude = max_val - min_val
    std_dev = np.std(data, ddof=1) # ddof=1 para desvio-padrão amostral
    print(f"\n--- Estatísticas para {name} ---")
    print(f"Média: {mean_val:.4f}")
    print(f"Mínimo: {min_val:.4f}")
    print(f"Máximo: {max_val:.4f}")
    print(f"Amplitude: {amplitude:.4f}")
    print(f"Desvio-padrão amostral: {std_dev:.4f}")
    return mean_val

# 1. Estatísticas para os dados originais
mean_original = calculate_stats(medidasOriginais, "Dados Originais")

# 2. Estatísticas para os dados arredondados
mean_arredondadas = calculate_stats(medidasArredondadas, "Dados Arredondados")

# 3. Estatísticas para os dados truncados
mean_truncadas = calculate_stats(medidasTruncadas, "Dados Truncados")

print("\n--- Erros da Média Arredondada em relação à Média Original ---")
print(f"Média de Referência (Original): {mean_original:.4f}")
print(f"Média Arredondada: {mean_arredondadas:.4f}")
print(f"Erro Absoluto da Média Arredondada: {Eabs(mean_original, mean_arredondadas):.6f}")
print(f"Erro Relativo da Média Arredondada: {Erel(mean_original, mean_arredondadas):.6f}")
print(f"Erro Percentual da Média Arredondada: {Eper(mean_original, mean_arredondadas):.6f}%")

print("\n--- Erros da Média Truncada em relação à Média Original ---")
print(f"Média de Referência (Original): {mean_original:.4f}")
print(f"Média Truncada: {mean_truncadas:.4f}")
print(f"Erro Absoluto da Média Truncada: {Eabs(mean_original, mean_truncadas):.6f}")
print(f"Erro Relativo da Média Truncada: {Erel(mean_original, mean_truncadas):.6f}")
print(f"Erro Percentual da Média Truncada: {Eper(mean_original, mean_truncadas):.6f}%")
```

A média original das medições é de 121.1200.
A média dos dados arredondados (121.1000) está muito próxima da original, com um erro absoluto de apenas 0.0200 e um erro percentual de 0.0165%. Isso sugere que o arredondamento para o inteiro mais próximo preserva bem a média nesta amostra.
A média dos dados truncados (120.7000) se afasta mais da média original, com um erro absoluto de 0.4200 e um erro percentual de 0.3468%. O truncamento tende a 'puxar' os valores para baixo, resultando em uma média sistematicamente menor.
Variabilidade (Desvio-padrão amostral):

O desvio-padrão amostral dos dados originais é de 0.8337.
Para os dados arredondados, o desvio-padrão é de 0.8756.
Para os dados truncados, o desvio-padrão é de 0.9487.
Ambas as formas de representação inteira aumentaram ligeiramente a estimativa do desvio-padrão em comparação com os dados originais. O truncamento resultou na maior alteração, indicando que a variabilidade da amostra é menos bem preservada quando os dados são simplesmente truncados para inteiros. O arredondamento, embora aumente o desvio-padrão, o faz em menor grau.

Conclusão:

Para esta pequena amostra, a representação inteira não é suficiente para preservar perfeitamente a média e a variabilidade. Embora o arredondamento preserve a média de forma razoavelmente boa, ele, juntamente com o truncamento, altera a estimativa da variabilidade (desvio-padrão). O truncamento, em particular, mostra um impacto mais significativo tanto na média quanto na variabilidade. Portanto, para análises que exijam alta precisão dessas estatísticas, a representação inteira não seria adequada. Não faremos interpretação médica dos valores.

# Parte C – vazão em um tubo de pequeno diâmetro

```python
import math

# Dados fornecidos
r_mm = 0.80 # mm
L = 0.20 # m
mu = 3.5 * (10**-3) # Pa s
delta_p = 1200 # Pa

# Conversão de unidades: raio de mm para m
r = r_mm / 1000 # m

# Fórmula de Hagen–Poiseuille
Q = (math.pi * (r**4) * delta_p) / (8 * mu * L)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# --- Parâmetros do Nanocateter ---
r_nm_base = 50       # raio em nm
L_um_base = 10       # comprimento em µm
mu_base = 3.5 * (10**-3) # viscosidade em Pa s
delta_p_base = 5000  # diferença de pressão em Pa

# --- Conversão para unidades SI (metros) ---
r_base = r_nm_base * 10**-9  # raio em metros
L_base = L_um_base * 10**-6  # comprimento em metros

# --- Função Hagen-Poiseuille ---
def calculate_Q(r_val, L_val, mu_val, delta_p_val):
    return (math.pi * (r_val**4) * delta_p_val) / (8 * mu_val * L_val)

# --- Vazão de Referência Q(r) ---
Q_ref_base = calculate_Q(r_base, L_base, mu_base, delta_p_base)
print(f"Vazão de Referência Q(r) para r={r_nm_base} nm: {Q_ref_base:.6e} m³/s\n")

# --- Variação do erro no raio (Delta_r) ---
delta_r_nm_values = np.arange(0.1, 5.1, 0.1) # ∆r de 0.1 a 5.0 nm

# Listas para armazenar os resultados
delta_r_nm_list = []
Q_r_minus_list = []
Q_r_ref_list = [] # Será o mesmo Q_ref_base para todos os delta_r
Q_r_plus_list = []

max_Eabs_list = []
max_Erel_list = []
max_Eper_list = []
linear_approx_Eper_list = []

for dr_nm in delta_r_nm_values:
    delta_r_nm_list.append(dr_nm)

    dr_m = dr_nm * 10**-9 # Converter delta_r para metros

    # Calcular r-, r+
    r_minus = r_base - dr_m
    r_plus = r_base + dr_m

    # Calcular vazões correspondentes
    Q_r_minus = calculate_Q(r_minus, L_base, mu_base, delta_p_base)
    Q_r_plus = calculate_Q(r_plus, L_base, mu_base, delta_p_base)

    Q_r_minus_list.append(Q_r_minus)
    Q_r_plus_list.append(Q_r_plus)
    Q_r_ref_list.append(Q_ref_base)

    # Calcular erros em relação a Q(r)
    err_abs_minus = Eabs(Q_ref_base, Q_r_minus)
    err_abs_plus = Eabs(Q_ref_base, Q_r_plus)
    max_Eabs_list.append(max(err_abs_minus, err_abs_plus))

    err_rel_minus = Erel(Q_ref_base, Q_r_minus)
    err_rel_plus = Erel(Q_ref_base, Q_r_plus)
    max_Erel_list.append(max(err_rel_minus, err_rel_plus))

    err_per_minus = Eper(Q_ref_base, Q_r_minus)
    err_per_plus = Eper(Q_ref_base, Q_r_plus)
    max_Eper_list.append(max(err_per_minus, err_per_plus))

    # Calcular a aproximação linear do erro percentual
    linear_approx_Eper = 4 * (dr_m / r_base) * 100
    linear_approx_Eper_list.append(linear_approx_Eper)

# --- Geração dos Gráficos ---

# 1. Gráfico de Q(r-), Q(r) e Q(r+) em função de ∆r
plt.figure(figsize=(12, 7))
plt.plot(delta_r_nm_list, Q_r_minus_list, label='Q(r - ∆r)', linestyle='--', marker='o', markersize=3)
plt.plot(delta_r_nm_list, Q_r_ref_list, label='Q(r) Referência', linestyle='-', color='gray', linewidth=2)
plt.plot(delta_r_nm_list, Q_r_plus_list, label='Q(r + ∆r)', linestyle='--', marker='x', markersize=3)
plt.title('Vazão (Q) em Função do Erro no Raio (∆r)')
plt.xlabel('Erro no Raio ∆r (nm)')
plt.ylabel('Vazão Q (m³/s)')
plt.legend()
plt.grid(True)
plt.ticklabel_format(style='sci', axis='y', scilimits=(0,0))
plt.show()

# 2. Gráfico do erro absoluto máximo de Q em função de ∆r
plt.figure(figsize=(12, 7))
plt.plot(delta_r_nm_list, max_Eabs_list, label='Erro Absoluto Máximo de Q', color='red', marker='o', markersize=3)
plt.title('Erro Absoluto Máximo da Vazão (Q) em Função do Erro no Raio (∆r)')
plt.xlabel('Erro no Raio ∆r (nm)')
plt.ylabel('Erro Absoluto Máximo de Q (m³/s)')
plt.legend()
plt.grid(True)
plt.ticklabel_format(style='sci', axis='y', scilimits=(0,0))
plt.show()

# 3. Gráfico do erro relativo máximo de Q em função de ∆r
plt.figure(figsize=(12, 7))
plt.plot(delta_r_nm_list, max_Erel_list, label='Erro Relativo Máximo de Q', color='green', marker='x', markersize=3)
plt.title('Erro Relativo Máximo da Vazão (Q) em Função do Erro no Raio (∆r)')
plt.xlabel('Erro no Raio ∆r (nm)')
plt.ylabel('Erro Relativo Máximo de Q')
plt.legend()
plt.grid(True)
plt.ticklabel_format(style='sci', axis='y', scilimits=(0,0))
plt.show()

# 4. Gráfico do erro percentual máximo de Q em função de ∆r
plt.figure(figsize=(12, 7))
plt.plot(delta_r_nm_list, max_Eper_list, label='Erro Percentual Máximo de Q', color='blue', marker='s', markersize=3)
plt.plot(delta_r_nm_list, linear_approx_Eper_list, label='Aproximação Linear (4 * ∆r/r * 100%)', color='orange', linestyle='--')
plt.title('Erro Percentual Máximo da Vazão (Q) em Função do Erro no Raio (∆r)')
plt.xlabel('Erro no Raio ∆r (nm)')
plt.ylabel('Erro Percentual Máximo de Q (%)')
plt.legend()
plt.grid(True)
plt.show()
```

A relação de Hagen–Poiseuille demonstra que a vazão Q é proporcional à quarta potência do raio r (Q ∝ r^4). Essa dependência forte implica que pequenas variações ou erros no raio são significativamente amplificados na vazão.

A aproximação linear ∆Q / Q ≈ 4∆r / r é uma ferramenta útil para estimar o erro relativo na vazão Q com base no erro relativo no raio r. Ela deriva da diferenciação da função Q(r) = C * r^4 (onde C é uma constante composta pelos demais termos): dQ/dr = 4 * C * r^3. Rearranjando, temos dQ/Q = 4 * dr/r ou, em termos de erros finitos, ∆Q/Q ≈ 4 * ∆r/r.

Para valores muito pequenos de ∆r, a aproximação linear é bastante precisa e se alinha bem com o erro percentual direto. Isso ocorre porque, quando ∆r é minúsculo, os termos de ordem superior na expansão binomial de (r ± ∆r)⁴ (como (∆r)², (∆r)³ e (∆r)⁴) tornam-se desprezíveis, e a relação linear domina.
À medida que ∆r aumenta, o erro percentual máximo de Q começa a se desviar um pouco da aproximação linear. Essa divergência ocorre porque os termos de ordem superior no erro do cálculo de Q(r-)/Q(r) e Q(r+)/Q(r) (que são (r-delta_r)^4 e (r+delta_r)^4) começam a ter um impacto maior. A aproximação linear subestima ligeiramente o erro real, especialmente para o Q(r-) (r-delta_r), pois a potência de 4 penaliza mais severamente a redução do raio.

```python
import numpy as np
import math
import matplotlib.pyplot as plt

# --- Re-using Nanocatheter Parameters from previous cells ---
# Ensure these are defined in the current scope for robustness
r_nm_base = 50       # raio em nm
L_um_base = 10       # comprimento em µm
mu_base = 3.5 * (10**-3) # viscosidade em Pa s
delta_p_base = 5000  # diferença de pressão em Pa

r_base_m = r_nm_base * 10**-9  # r in meters (float64 by default)
L_base_m = L_um_base * 10**-6  # L in meters (float64 by default)

# --- Function for Q calculation (re-defined to accept dtype) ---
def calculate_Q_with_dtype(r_val, L_val, mu_val, delta_p_val, dtype):
    r_val = dtype(r_val)
    L_val = dtype(L_val)
    mu_val = dtype(mu_val)
    delta_p_val = dtype(delta_p_val)
    pi_val = dtype(np.pi)
    return (pi_val * (r_val**4) * delta_p_val) / (dtype(8) * mu_val * L_val)

# --- Calculations for float64 ---
r_float64 = np.float64(r_base_m)
Q_float64 = calculate_Q_with_dtype(r_base_m, L_base_m, mu_base, delta_p_base, np.float64)
r4_float64 = r_float64**4

# --- Calculations for float32 ---
r_float32 = np.float32(r_base_m)
Q_float32 = calculate_Q_with_dtype(r_base_m, L_base_m, mu_base, delta_p_base, np.float32)
r4_float32 = r_float32**4

# --- Record and Print Results ---
print("--- Comparação de Precisão (float32 vs float64) ---\n")

print(f"r (original, m): {r_base_m:.2e}")
print(f"L (original, m): {L_base_m:.2e}")
print(f"μ (original, Pa s): {mu_base:.2e}")
print(f"∆p (original, Pa): {delta_p_base:.0f}")

print(f"")
print(f"r^4 (float64): {r4_float64:.6e}")
print(f"r^4 (float32): {r4_float32:.6e}")
print(f"Diferença Absoluta (r^4): {Eabs(r4_float64, r4_float32):.2e}")
print(f"Diferença Relativa (r^4): {Erel(r4_float64, r4_float32):.2e}")

print(f"")
print(f"Q (float64): {Q_float64:.6e} m³/s")
print(f"Q (float32): {Q_float32:.6e} m³/s")
print(f"Diferença Absoluta (Q): {Eabs(Q_float64, Q_float32):.2e}")
print(f"Diferença Relativa (Q): {Erel(Q_float64, Q_float32):.2e}")

print(f"")
print(f"np.finfo(float64).eps: {np.finfo(np.float64).eps:.2e}")
print(f"np.finfo(float32).eps: {np.finfo(np.float32).eps:.2e}")

# --- Cálculos para δ1 e δ2 ---
print("\n--- Comparação de δ1 e δ2 para perda de significância ---\n")

# Use the original Q_ref_base from previous calculations
# Re-define calculate_Q for simplicity in this section, always using float64
def calculate_Q_float64_single_arg(r_val_m):
    return (math.pi * (r_val_m**4) * delta_p_base) / (8 * mu_base * L_base_m)

# Values for delta_r_over_r
delta_r_over_r_values = np.logspace(-1, -16, 50) # From 0.1 down to 1e-16

delta1_values = []
delta2_values = []

for dr_over_r in delta_r_over_r_values:
    # Ensure all calculations are done in float64 to clearly observe cancellation effects
    current_r = np.float64(r_base_m)
    current_dr_over_r = np.float64(dr_over_r)
    delta_r = current_r * current_dr_over_r

    # δ1 = (Q(r + ∆r) − Q(r)) / Q(r)
    Q_r_plus_dr = calculate_Q_float64_single_arg(current_r + delta_r)
    delta1 = (Q_r_plus_dr - calculate_Q_float64_single_arg(current_r)) / calculate_Q_float64_single_arg(current_r)
    delta1_values.append(delta1)

    # δ2 = (1 + ∆r / r)^4 - 1
    delta2 = (np.float64(1) + current_dr_over_r)**4 - np.float64(1)
    delta2_values.append(delta2)

# --- Plotting the comparison ---
plt.figure(figsize=(12, 7))
plt.loglog(delta_r_over_r_values, delta1_values, 'o-', label='$\delta_1 = (Q(r + \Delta r) - Q(r)) / Q(r)$ (cálculo direto)')
plt.loglog(delta_r_over_r_values, delta2_values, 'x--', label='$\delta_2 = (1 + \Delta r / r)^4 - 1$ (fórmula rearranjada)')
plt.xlabel('$\Delta r / r$')
plt.ylabel('Variação Relativa na Vazão')
plt.title('Comparação de $\delta_1$ e $\delta_2$ em Função de $\Delta r / r$')
plt.legend()
plt.grid(True, which="both", ls="-")
plt.show()
```

Para valores grandes de $\Delta r / r$$\Delta r / r$ (à esquerda do gráfico), ambas as expressões fornecem resultados muito próximos. Isso indica que a aritmética de ponto flutuante tem precisão suficiente para lidar com os valores envolvidos.

No entanto, para valores progressivamente menores de $\Delta r / r$$\Delta r / r$ (movendo-se para a direita no gráfico), a expressão $\delta_1$$\delta_1$ começa a divergir da expressão $\delta_2$$\delta_2$.

A expressão $\delta_1 = (Q(r + \Delta r) - Q(r)) / Q(r)$$\delta_1 = (Q(r + \Delta r) - Q(r)) / Q(r)$ envolve a subtração de dois números muito próximos (Q(r + $\Delta$$\Delta$r) e Q(r)) quando $\Delta r$$\Delta r$ é pequeno. Esta operação, conhecida como cancelamento catastrófico, leva à perda de dígitos significativos, pois os bits mais significativos dos números se anulam, deixando apenas os bits menos precisos do resultado.

Por outro lado, a expressão $\delta_2 = (1 + \Delta r / r)^4 - 1$$\delta_2 = (1 + \Delta r / r)^4 - 1$ é algebricamente equivalente a $\delta_1$$\delta_1$, mas sua forma evita a subtração direta de termos muito próximos. Ao invés disso, ela calcula uma pequena quantidade elevada à quarta potência e depois subtrai 1, mantendo a precisão por mais tempo.

A divergência entre $\delta_1$$\delta_1$ e $\delta_2$$\delta_2$ no lado direito do gráfico demonstra claramente a perda de significância por cancelamento na expressão $\delta_1$$\delta_1$ quando $\Delta r / r$$\Delta r / r$ se torna menor que a precisão da máquina (np.finfo(float64).eps). Isso ressalta a importância de escolher a formulação algorítmica correta em cálculos de ponto flutuante para evitar erros numéricos.

A redução da escala de um sistema, como no caso do nanocateter, aumenta drasticamente a razão superfície-volume. Este fenômeno faz com que os efeitos de superfície e os efeitos viscosos (ou de parede) se tornem relativamente mais dominantes em comparação com as forças de volume. Em fluidodinâmica, isso significa que a influência do atrito do fluido com as paredes do canal e a viscosidade intrínseca do fluido adquirem uma importância muito maior no comportamento do escoamento, moldando significativamente a vazão e a dinâmica interna do sistema. Contudo, é crucial entender que, embora um cálculo possa apresentar alta precisão aritmética, com muitas casas decimais, o resultado pode ser fisicamente inadequado se o modelo matemático subjacente não capturar de forma fiel os fenômenos físicos dominantes na escala de interesse. A precisão dos números não garante, por si só, a validade física do modelo ou a representatividade da realidade.

```python
import matplotlib.pyplot as plt
import numpy as np
import math
from matplotlib.gridspec import GridSpec
from matplotlib import animation
from matplotlib.patches import Circle

# --- Nanocatheter Parameters (retrieved from previous cells) ---
r_nm_base = 50       # reference radius in nm
L_um_base = 10       # length in µm
mu_base = 3.5 * (10**-3) # viscosity in Pa s
delta_p_base = 5000  # pressure difference in Pa

# Convert to SI units (meters)
r_ref_m = r_nm_base * 10**-9
L_base_m = L_um_base * 10**-6

# --- Function Hagen-Poiseuille (from previous cells) ---
def calculate_Q(r_val, L_val, mu_val, delta_p_val):
    return (math.pi * (r_val**4) * delta_p_val) / (8 * mu_val * L_val)

# --- Error functions (from previous cells) ---
def Eabs(x, xn):
  return abs(x - xn)

def Erel(x, xn):
  # Handle division by zero for relative error if x is 0
  if x == 0:
      return 0 if xn == 0 else float('inf') * np.sign(xn)
  return (abs(x - xn)/abs(x))

def Eper(x, xn):
  return 100*Erel(x, xn)

# --- Calculate reference Q at r = 50 nm ---
Q_ref_50nm = calculate_Q(r_ref_m, L_base_m, mu_base, delta_p_base)

# --- Generate radius sequence for animation ---
r_nm_start = 45
r_nm_end = 55
num_frames = 50 # 30 to 60 frames

r_nm_animation_values = np.linspace(r_nm_start, r_nm_end, num_frames)
r_m_animation_values = r_nm_animation_values * 10**-9

# --- Setup the figure for animation ---
fig = plt.figure(figsize=(20, 12), dpi=100)
gs = GridSpec(nrows=2, ncols=2, figure=fig, height_ratios=[1, 1])

# Subplot for channel cross-section
ax_cross_section = fig.add_subplot(gs[0, 0])
# Subplot for text display
ax_text = fig.add_subplot(gs[0, 1])
# Subplot for progressive Q vs r curve
ax_plot = fig.add_subplot(gs[1, :]) # Spans both columns in the second row

# Initialize lists for progressive plot (global for update to append)
r_history_nm = []
Q_history = []

# Pre-calculate global min/max for Q plot limits to keep them fixed
Q_global_min_for_plot = calculate_Q(r_nm_start * 10**-9, L_base_m, mu_base, delta_p_base)
Q_global_max_for_plot = calculate_Q(r_nm_end * 10**-9, L_base_m, mu_base, delta_p_base)

# --- Update function for animation frames ---
def update(i):
    # Clear axes for each frame
    ax_cross_section.clear()
    ax_text.clear()
    ax_plot.clear()

    # Get current values
    r_current_nm = r_nm_animation_values[i]
    r_current_m = r_m_animation_values[i]

    # Calculate Q and error
    Q_current = calculate_Q(r_current_m, L_base_m, mu_base, delta_p_base)
    error_percent = Eper(Q_ref_50nm, Q_current)

    # Update history lists for progressive plot
    r_history_nm.append(r_current_nm)
    Q_history.append(Q_current)

    # --- Plot cross-section ---
    ax_cross_section.set_aspect('equal', adjustable='box')
    # Fixed limits for cross-section to prevent jumping
    ax_cross_section.set_xlim(-r_nm_end * 1.1, r_nm_end * 1.1)
    ax_cross_section.set_ylim(-r_nm_end * 1.1, r_nm_end * 1.1)
    ax_cross_section.set_title(f'Seção Transversal do Canal (Raio: {r_current_nm:.2f} nm)', fontsize=14)
    ax_cross_section.axis('off') # Hide axes for a cleaner circle representation
    circle = Circle((0, 0), r_current_nm, color='skyblue', alpha=0.6, edgecolor='blue', linewidth=2)
    ax_cross_section.add_patch(circle)

    # --- Display text ---
    ax_text.axis('off')
    ax_text.text(0.05, 0.9, f'Raio Atual: {r_current_nm:.2f} nm', transform=ax_text.transAxes, fontsize=12)
    ax_text.text(0.05, 0.7, f'Vazão Calculada: {Q_current:.2e} m³/s', transform=ax_text.transAxes, fontsize=12)
    ax_text.text(0.05, 0.5, f'Erro % (ref. 50 nm): {error_percent:.2f}%', transform=ax_text.transAxes, fontsize=12,
                 color='red' if abs(error_percent) > 0.1 else 'green')
    ax_text.set_title('Dados do Quadro Atual', fontsize=14)

    # --- Plot progressive Q vs r curve ---
    ax_plot.plot(r_history_nm, Q_history, color='blue', linewidth=2, marker='o', markersize=4)
    ax_plot.plot(r_current_nm, Q_current, marker='X', markersize=10, color='red', label='Ponto Atual') # Highlight current point

    # Set fixed limits for the progressive plot for better visualization
    ax_plot.set_xlim(r_nm_start - 1, r_nm_end + 1)
    ax_plot.set_ylim(Q_global_min_for_plot * 0.9, Q_global_max_for_plot * 1.1) # Fixed global limits

    ax_plot.set_xlabel('Raio r (nm)', fontsize=12)
    ax_plot.set_ylabel('Vazão Q (m³/s)', fontsize=12)
    ax_plot.set_title('Curva de Vazão (Q) vs Raio (r) (Progressivo)', fontsize=14)
    ax_plot.ticklabel_format(style='sci', axis='y', scilimits=(0,0))
    ax_plot.grid(True)
    ax_plot.legend()

    return []

# Create the animation
anim = animation.FuncAnimation(fig, update, frames=num_frames, interval=200, blit=False) # interval in ms

# Save the animation as a GIF
gif_filename = 'nanocatheter_flow_animation.gif'
# Use PillowWriter for GIF. fps controls frames per second (e.g., 10 frames per second)
anim.save(gif_filename, writer='pillow', fps=10)

plt.close(fig) # Close the plot to prevent it from displaying static frame

print(f"GIF '{gif_filename}' gerado com sucesso!")
```
